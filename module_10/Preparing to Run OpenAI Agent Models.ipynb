# Preparing to Run OpenAI Agent Models

## Step 1: Activate Environment and Install Package

Open a terminal and activate the course Conda environment:
conda activate AI500Env

Install the OpenAI Agents package: 
pip install openai-agents

## Step 2 Clone OpenAI Agents SDK from github.

Open modele_10 directory

git clone https://github.com/openai/openai-agents-python

## Step 3 Copy Your .env File (with API Key)

go to module_10/openai-agents-python

cp ../env .

add the following code to the main.py of the agent you want to run
