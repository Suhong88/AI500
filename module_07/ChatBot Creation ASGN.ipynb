# Take Home Assignment: Create your own orderbot

Design and implement a conversational app using OpenAI Model that takes, validates, summarizes, and confirms customer orders for a chosen business (e.g., coffee shop, smoothie bar, bookstore, salon, food truck). Please test your bot and make sure it can answer questions correctly. 

Testing Checklist

- Can the bot reject an invalid size and ask the right question?

- Does it capture quantities and modifiers reliably?

- Are totals correct for 3+ items with multiple modifiers?

- Does the summary exactly match the cart?

- Can a user change or remove an item mid-flow?

- Are orders saved after confirmation?

Please submit your completed Jupyter notebook, a html version of the notbook, along with a Word document that explains how your OrderBot functions. Include several screenshots demonstrating how you tested the chatbot to ensure it produces the desired responses.
