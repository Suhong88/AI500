# Take Home Assignment: Image Creation

Choose one image generation model from the following options:

- GPT-5 (Image generation feature)
- Open-source Stable Diffusion XL (SDXL)
- Open-source Stable Diffusion 3 Medium
- Stability AI Web Interface


Create five images on topics or scenes of your choice. Experiment with different prompts or refine your original prompts to improve the results. Save your generated images.

Write a short report (1–2 pages, Word document) that includes:

- The prompts you used for each image.
- the images you created.
- How you refined or modified your prompts to improve image quality.
- The challenges you encountered during the process (e.g., prompt sensitivity, rendering speed, model limits).
- Your reflections on the limitations and creative potential of the model you used.

Please submit your completed Jupyter notebook (if appplicable) along with a Word document that discusses the above points.
