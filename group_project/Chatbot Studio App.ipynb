# Group Project: Chatbot Studio (No-Code Option)

**Chatbot Studio** lets a team build a customer-service chatbot with **tool calling** for almost any business, without
writing code: a restaurant or shop (orders), a salon, clinic or tutoring center (appointments), a rental business
(bookings and quotes), a help desk (answers from documents and support tickets), and more. It extends the OrderBot
Builder from Mini-project #4.

**How it works.** You describe your business in plain English. An advanced OpenAI model (`gpt-6.2-astra`) drafts the
whole bot for you. You then check and improve every part, in forms or by asking for changes in plain English. The bot
acts only through **ready-made tools**; the AI fills in their settings, and the app's code runs them and enforces the
rules (prices come from your tables, nothing is saved before the customer says yes, a full time slot cannot be booked).

| Tool type | What it does | Good for |
|---|---|---|
| 🔎 Look up | Searches one of your data tables | menus, products, rooms, courses |
| 🧮 Calculate | Works out a price: price x quantity, extras, fees, tax | order totals, quotes |
| 📝 Collect and save | Gathers details, shows a summary, saves after a yes | orders, bookings, tickets |
| 📅 Check availability | Lists open slots; a booking uses one up | appointments, seats |
| 📚 Answer from documents | Finds the passages that answer a question (RAG) | policies, FAQ |
| 🙋 Hand off to a person | Records a request for staff to follow up | complaints, special requests |

**The tabs:** 1. Start (draft or change the bot) · 2. Instructions and rules · 3. Data (tables and documents) ·
4. Tools (read, test and change each tool) · 5. Test (10-20 test conversations, checked by AI and by you) ·
6. Chat · 7. Results (the file you submit).

**Before you start:** your `OPENAI_API_KEY` must be in the `.env` file in the parent folder (the same file as in
Module 7). A draft or a change costs about 10-25 cents; a run of 12 tests with gpt-5.4-mini about 10 cents.
